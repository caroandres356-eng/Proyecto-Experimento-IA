# 🧠 Proyecto: Support Vector Machines (SVM) vs Red Neuronal MLP

Este notebook implementa el proyecto final de clasificación utilizando el dataset de enfermedades cardíacas (`heart.csv`).

### Pasos del Proyecto:
1. **Carga y Preprocesamiento**: Cargar los datos y aplicar un escalador estándar.
2. **Entrenamiento y Ajuste de SVM (Support Vector Machines)**.
3. **Entrenamiento y Ajuste de MLP (Multilayer Perceptron)**.
4. **Análisis Comparativo y Conclusiones**.

```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Agregar el directorio raíz al path para poder importar módulos compartidos
sys.path.append(os.path.abspath('..'))
from shared.data_loader import load_and_preprocess_data, get_default_data_path

print("¡Librerías importadas con éxito!")
```

## 1. Carga y Preparación de Datos

```python
# Cargar datos escalados (esencial para SVM y MLP, ya que son muy sensibles a la escala de las variables)
X_train, X_test, y_train, y_test, scaler = load_and_preprocess_data(scale_features=True)
print(f"Entrenamiento: {X_train.shape[0]} muestras")
print(f"Prueba: {X_test.shape[0]} muestras")
```

## 2. Modelo 1: Support Vector Machines (SVM)

```python
from sklearn.svm import SVC
from sklearn.metrics import classification_report, confusion_matrix

# Entrenar modelo SVM con kernel RBF y regularización C=1.0
svm = SVC(C=1.0, kernel='rbf', random_state=42)
svm.fit(X_train, y_train)
y_pred_svm = svm.predict(X_test)

print(classification_report(y_test, y_pred_svm))
```

## 3. Modelo 2: Red Neuronal Perceptrón Multicapa (MLP)

```python
from sklearn.neural_network import MLPClassifier

# Entrenar la red neuronal MLP con la mejor arquitectura encontrada durante el ajuste de hiperparámetros
mlp = MLPClassifier(hidden_layer_sizes=(50,25), activation='relu', max_iter=500, random_state=42, early_stopping=True, validation_fraction=0.1)
mlp.fit(X_train, y_train)
y_pred_mlp = mlp.predict(X_test)

print(classification_report(y_test, y_pred_mlp))
```
