```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Agregar el directorio raíz al path para poder importar módulos compartidos
sys.path.append(os.path.abspath('..'))
from shared.data_loader import load_and_preprocess_data, get_default_data_path

print("¡Librerías importadas con éxito!")
```

## 1. Análisis Exploratorio de Datos (EDA)

```python
# Cargar el dataset crudo para explorar
path = get_default_data_path()
df = pd.read_csv(path)

# Mostrar primeras 5 filas
df.head()
```

```python
# Información general del dataset (tipos de datos, no-nulos)
df.info()
```

```python
# Distribución de la variable objetivo (target)
plt.figure(figsize=(6, 4))
sns.countplot(x='target', data=df, palette='Blues')
plt.title('Distribución de Enfermedad Cardíaca (0 = Sano, 1 = Enfermo)')
plt.xlabel('Diagnóstico')
plt.ylabel('Cantidad')
plt.show()
```

## 2. Preparación y Carga de Datos Normalizados

```python
# Cargar datos divididos y escalados
X_train, X_test, y_train, y_test, scaler = load_and_preprocess_data(scale_features=True)
print(f"Entrenamiento: {X_train.shape[0]} muestras")
print(f"Prueba: {X_test.shape[0]} muestras")
```

## 3. Modelo 1: KNN

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix

# Probar con un valor de K por defecto
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
y_pred_knn = knn.predict(X_test)

print(classification_report(y_test, y_pred_knn))
```

## 4. Modelo 2: Árbol de Decisión

```python
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier(max_depth=3, random_state=42)
dt.fit(X_train, y_train)
y_pred_dt = dt.predict(X_test)

print(classification_report(y_test, y_pred_dt))
```
